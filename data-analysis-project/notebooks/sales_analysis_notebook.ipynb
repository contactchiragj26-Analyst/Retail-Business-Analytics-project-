# Retail Business Analysis

This notebook demonstrates a retail analytics workflow using a realistic sales dataset. It focuses on revenue performance, category profitability, channel efficiency, and operational risk indicators such as returns.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/sales_data.csv', parse_dates=['order_date'])
df.head()
```

```python
monthly = (
    df.assign(month=df['order_date'].dt.to_period('M').astype(str))
    .groupby('month', as_index=False)['sales']
    .sum()
    .set_index('month')
)
monthly.plot(kind='line', marker='o', figsize=(10, 5))
plt.title('Monthly Sales Trend')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
kpis = {
    'total_revenue': df['sales'].sum(),
    'total_profit': df['profit'].sum(),
    'average_order_value': df['sales'].mean(),
    'top_region': df.groupby('region')['sales'].sum().idxmax(),
    'return_rate': (df['returns'].mean() * 100),
}
kpis
```
